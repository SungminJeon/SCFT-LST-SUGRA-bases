# 6D SUGRA Base Catalog — tutorial

Explore the catalog with a few lines of Python using `sugra.py`.
Run the cells top to bottom. Edit `ROOT` to point at your data folder.

```python
import sugra

ROOT = "catalog"          # <-- folder that contains T2_2/, T3_3/, ...
cat = sugra.Catalog(ROOT)
cat.tensors()             # which T_H are available
```

## What combos exist at a given T_H?

```python
combos = cat.combos(8)
print(len(combos), "combos at T_H=8")
combos[:10]
```

## Load the bases for one combo

Each `Base` carries the physics quantities and the intersection-form matrix.

```python
bases = cat.load(8, "e6+su3")
print(len(bases), "bases")
b = bases[0]
print(b)
print("tags    :", b.tags)
print("physics :", b.physics)
print("T_min   :", b.t_min())
b.IF                      # numpy intersection-form matrix
```

## Filtering when you don't know the combo

- `externals=[...]` — exact multiset (order doesn't matter)
- `contains="e8"`   — at least one such external
- `t_min=...`        — by T_min

```python
hits = cat.find(t_h=8, externals=["e6", "su3"])
print("e6+su3 at T_H=8 :", len(hits))

with_e8 = cat.find(t_h=8, contains="e8")
print("contains e8     :", len(with_e8))
```

## No-code route: the CSV index

Build it once with `python build_index.py <ROOT>`, then filter with pandas
(or open `summary.csv` in a spreadsheet).

```python
import pandas as pd
df = pd.read_csv(f"{ROOT}/index.csv")
print("total bases:", len(df))
df[(df.T_H == 8) & (df.combo == "e6+su3")].head()
```

```python
# how many bases contain an e8 external, per T_H?
df[df.combo.str.contains("e8")].groupby("T_H").size()
```
